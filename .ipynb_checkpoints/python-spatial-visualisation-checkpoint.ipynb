# Картограммы (choropleth)

При создании такого типа карт, цветом или штриховкой будет отображаться значение рассматриваемой переменной. 
Примеры можно посмотреть [здесь](https://baltti.github.io/gis-itmo/visualisation.html#%D0%BA%D0%B0%D1%80%D1%82%D0%BE%D0%B3%D1%80%D0%B0%D0%BC%D0%BC%D0%B0-choropleth).

Главная рекомендация при их построении состоит в том, что следует использовать **относительные показатели переменных, а не абсолютные.**

Кроме того, важными являются следующие вопросы: 
- классификации численной переменной; 
- разбиения ее на интервалы; 
- подбор нужного количества интервалов, 

так как это будет напрямую влиять на то, как информация воспринимается зрителем.

````python
import seaborn as sn
import pandas as pd
import geopandas as gpd
import pysal
import numpy
import matplotlib.pyplot as plt
````

## Ознакомительный анализ датасета

Загрузим наши данные из файла.

В качестве исходных данных мы воспользуемся сведениями о населении по муниципальным образованиям Санкт-Петербурга. Эти данные взяты из [сведения Росстата](https://petrostat.gks.ru/folder/27595) по состоянию на 1 января 2020 года.

````python
mo = gpd.read_file("https://raw.githubusercontent.com/baltti/python-spatial-itmo/main/datasets/mo.geojson")

mo.head()
````

Рассчитаем плотность населения на основе данных о количества населения и площади муниципальных образований.

Это нужно, чтобы мы могли сразу строить карту по относительному показателю, а не по абсолютным значениям.

````python
mo["density"] = mo["population"]/mo["area_km2"]

mo.head()
````

Далее посмотрим, как распределена интересующая нас переменная. 

````python
ax = sn.histplot(mo["density"], bins=5)

sn.rugplot(mo["density"], height=0.05, color="red", ax=ax);
````

Из графика видно, что данные немного смещены влево.

Это подтверждается и статистическими показателями: так среднее значение  в нашем случае больше медианного (50 % на результатах представленных ниже).

Этот факт будет оказывать влияние на классификацию данных.

````python
mo["density"].describe()
````

## Классификация



````python
# pip install mapclassify
````

````python
import mapclassify
````

### Равные интервалы

Один из самых простых видов классификации, который заключается в том, что весь интервал разбивается на нужное количество равных отрезков.

````python
equal_interval = mapclassify.EqualInterval(mo["density"], k=5)

equal_interval
````

### Квантили

При использовании такой классификации объекты разбиваются на группы так, чтобы в каждой группе было одинаковое число объектов.

````python
quantiles = mapclassify.Quantiles(mo["density"], k=5)

quantiles
````

Сравним размер интервалов для каждой группы.

````python
quantiles.bins[1:] - quantiles.bins[:-1]
````

Главная сложность в использовании такого метода классификации в том, что размер интервалов может быть очень разным, что приводит к сложности интерпретации результатов.

### Стандартное отклонение

В этом случае классы рассчитываются по величине стандартного отклонения относительно медианы.

Например, в нашем случае при использовании 5 интервалов мы получим интервалы $\left(\overline{x}-2s\right)$, $\left(\overline{x}-s\right)$, $\left(\overline{x}+s\right)$, $\left(\overline{x}+2s\right)$.
Все, что не попало в эти интервалы будет отнесено в нижний или верхний интервал соответственно.

````python
stand_dev = mapclassify.StdMean(mo["density"])

stand_dev
````

Такую классификацию хорошо использовать, если данные нормально распределены.

### Естественные интервалы (класификация Фишера-Дженкса)

Эта классификация считается одной из оптимальных.

Метод построен на том, что внутри классов минимизируются суммы отклонений от медиан классов.

````python
fisher_jenks = mapclassify.FisherJenks(mo["density"], k=5)

fisher_jenks
````

### Сравнение результатов классификации

Проанализируем разницу между группами, полученную различными методами классификации.
Для этого сравним абсолютные отклонения внутри классов от их медиан (absolute deviation around class medians - ADCM). Это позволит увидеть, насколько полученные группы "компактны".


````python
# Bunch classifier objects
class5 = equal_interval, quantiles, stand_dev, fisher_jenks

# Collect ADCM for each classifier
fits = numpy.array([c.adcm for c in class5])

# Convert ADCM scores to a DataFrame
adcms = pd.DataFrame(fits)

# Add classifier names
adcms["classifier"] = [c.name for c in class5]

# Add column names to the ADCM
adcms.columns = ["ADCM", "Classifier"]
ax = sn.barplot(y="Classifier", 
                x="ADCM", 
                data=adcms, 
                palette="Pastel1")
````

Чем меньше полученное значение, тем более компактными являются группы, тем лучше полученная классификация.

Кроме сравнения компактности классов, еще интересно посмотреть, к какому классу были отнесены одни и те же объекты при использовании различных методов.

Добавим номера классов в отдельные колонки нашего датасета.

````python
mo["Quantiles"] = quantiles.yb
mo["Equal Interval"] = equal_interval.yb
mo["Mean-Standard Deviation"] = stand_dev.yb
mo["Fisher-Jenks"] = fisher_jenks.yb
````

Отобразим полученные результаты на тепловой карте.

````python
f, ax = plt.subplots(1, figsize=(15, 5))
sn.heatmap(
           mo.set_index("NAME")
           .sort_values("density")[
                            [
                             "Fisher-Jenks",
                             "Equal Interval",
                             "Quantiles",
                             "Mean-Standard Deviation",
                            ]
                           ]
            .T,
            cmap="YlGn",
            cbar=False,
            ax=ax,
           )

ax.set_xlabel("Муниципальный округ")
````

На полученном графике методы классификации представлены по строкам, а объекты - по столбцам. Цвет зависит от группы, в которую объект был отнесен.

Также сравним, сколько объектов было отнесено к той или иной группе в различных методах классификации.

````python
pd.DataFrame(
             {c.name: c.counts for c in class5},
             index=["Class-{}".format(i) for i in range(5)],
            )
````

## Построение карты

````python
ax = mo.plot(
             column="density",  # Data to plot
             scheme="Quantiles",  # Classification scheme
             cmap="YlGn",  # Color palette
             legend=True,  # Add legend
             legend_kwds={"fmt": "{:.0f}"},  # Remove decimals in legend
             figsize=(10, 15)
            )
ax.set_axis_off();
````

Фактически при построении картограммы для объекта ```GeoDataFrame``` используется один из методов класса ```.plot()```, в аргументах которого нужно указать переменную для отображения, метод классификации (эти методы могут быть указаны напрямую из библиотеки ```mapclassify```, как было рассмотрено в предыдущих разделах) и цветовую палитру.



При построении карты кроме классификации важным еще является вопрос подбора цветовой палитры.

Существует три основных типа палитр: последовательные, расходящиеся и качественные (sequential, diverging, qualitative).

Последовательные палитры - градиенты, которые рекомендуется использовать для непрерывных числовых данных. Они позволяют довольно однозначное трактование таких значений.

Так, например, на рисунке выше муниципальные образования с низкой плотностью населения показаны светлыми оттенками палитры, а с высокой - темными.

Такие палитры также могут быть и оттенками одного цвета.

````python
ax = mo.plot(
             column="density",  # Data to plot
             scheme="Quantiles",  # Classification scheme
             cmap="Blues",  # Color palette
             edgecolor="k",  # Borderline color
             linewidth=0.1,  # Borderline width
             legend=True,  # Add legend
             legend_kwds={"fmt": "{:.0f}"},  # Remove decimals in legend
             figsize=(10, 15)
            )
ax.set_axis_off();
````

Качественные палитры рекомендуется использовать для качественных или дискретных переменных, так как они акцентируют внимание на разнице между группами объектов.

````python
ax = mo.plot("ADMIN_L5", categorical=True,
             legend=True, figsize=(10, 15))
ax.set_axis_off();
````

Расходящиеся цветовые палитры тоже подходят для непрерывных числовых данных, но у них есть определенная особенность: в таких палитрах акцентируется внимание на середине всего распределения чисел и минимальными с максимальными значениями.

Например, такие палитры часто используются для температур, когда нужно показать положительные и отрицательные значения, а также ноль.

````python
import folium

m = mo.explore(
     column="density", # make choropleth based on "BoroName" column
     scheme="fisherjenks",
     cmap="Blues",
     tooltip=("NAME", "population", "density"), # show "BoroName" value in tooltip (on hover)
     popup=True, # show all values in popup (on click)
     style_kwds=dict(color="black") # use black outline
    )

folium.TileLayer('Stamen Toner', control=True).add_to(m)  # use folium to add alternative tiles
folium.LayerControl().add_to(m)  # use folium to add layer control

m  # show map
````

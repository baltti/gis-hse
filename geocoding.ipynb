```python
import geodatasets
import matplotlib.pyplot as plt
import geopandas as gpd
from geopandas import tools 
```

```python
boros = gpd.read_file(geodatasets.get_path("nybb"))

boros.BoroName
```

```python
boro_locations = gpd.tools.geocode(boros.BoroName)

boro_locations
```

```python
fig, ax = plt.subplots()

boros.to_crs("EPSG:4326").plot(ax=ax, color="white", edgecolor="black");

boro_locations.plot(ax=ax, color="red");
```
